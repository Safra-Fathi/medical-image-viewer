```python
IMG_SIZE = 256
BATCH_SIZE = 8

from PIL import Image
import numpy as np
import tensorflow as tf


def load_image_mask(image_path, mask_path):

    # Convert TensorFlow tensors to normal Python paths
    image_path = image_path.numpy().decode("utf-8")
    mask_path = mask_path.numpy().decode("utf-8")

    # Load MRI image
    image = Image.open(image_path).convert("RGB")
    image = np.array(image).astype(np.float32) / 255.0

    # Load tumor mask
    mask = Image.open(mask_path).convert("L")
    mask = np.array(mask).astype(np.float32)

    # Convert 0/255 → 0/1
    mask = (mask > 0).astype(np.float32)

    # Add channel dimension
    mask = np.expand_dims(mask, axis=-1)

    return image, mask


def load_image_mask_tf(image_path, mask_path):

    image, mask = tf.py_function(
        func=load_image_mask,
        inp=[image_path, mask_path],
        Tout=[tf.float32, tf.float32]
    )

    image.set_shape([IMG_SIZE, IMG_SIZE, 3])
    mask.set_shape([IMG_SIZE, IMG_SIZE, 1])

    return image, mask
```

```python
def create_dataset(pairs, batch_size=8, shuffle=False):

    image_paths = [p[0] for p in pairs]
    mask_paths = [p[1] for p in pairs]

    dataset = tf.data.Dataset.from_tensor_slices(
        (image_paths, mask_paths)
    )

    dataset = dataset.map(
        load_image_mask_tf,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if shuffle:
        dataset = dataset.shuffle(500)

    dataset = dataset.batch(batch_size)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset


train_ds = create_dataset(train_pairs, BATCH_SIZE, shuffle=True)
val_ds = create_dataset(val_pairs, BATCH_SIZE)
test_ds = create_dataset(test_pairs, BATCH_SIZE)

print("Datasets created successfully!")
```

# Create the U-Net model

from tensorflow.keras import layers, Model


def conv_block_small(x, filters):
    x = layers.Conv2D(
        filters, 3, padding="same", activation="relu"
    )(x)
    x = layers.BatchNormalization()(x)

    x = layers.Conv2D(
        filters, 3, padding="same", activation="relu"
    )(x)
    x = layers.BatchNormalization()(x)

    return x


def build_small_unet(input_shape=(256, 256, 3)):

    inputs = layers.Input(input_shape)

    # Encoder
    c1 = conv_block_small(inputs, 16)
    p1 = layers.MaxPooling2D(2)(c1)

    c2 = conv_block_small(p1, 32)
    p2 = layers.MaxPooling2D(2)(c2)

    c3 = conv_block_small(p2, 64)
    p3 = layers.MaxPooling2D(2)(c3)

    c4 = conv_block_small(p3, 128)
    p4 = layers.MaxPooling2D(2)(c4)

    # Bottleneck
    c5 = conv_block_small(p4, 256)

    # Decoder
    u6 = layers.Conv2DTranspose(
        128, 2, strides=2, padding="same"
    )(c5)
    u6 = layers.concatenate([u6, c4])
    c6 = conv_block_small(u6, 128)

    u7 = layers.Conv2DTranspose(
        64, 2, strides=2, padding="same"
    )(c6)
    u7 = layers.concatenate([u7, c3])
    c7 = conv_block_small(u7, 64)

    u8 = layers.Conv2DTranspose(
        32, 2, strides=2, padding="same"
    )(c7)
    u8 = layers.concatenate([u8, c2])
    c8 = conv_block_small(u8, 32)

    u9 = layers.Conv2DTranspose(
        16, 2, strides=2, padding="same"
    )(c8)
    u9 = layers.concatenate([u9, c1])
    c9 = conv_block_small(u9, 16)

    outputs = layers.Conv2D(
        1, 1, activation="sigmoid"
    )(c9)

    return Model(inputs, outputs)


model = build_small_unet()

model.summary()

# Add Dice loss

```python
def dice_coefficient(y_true, y_pred, smooth=1e-6):

    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)

    y_true = tf.reshape(y_true, [-1])
    y_pred = tf.reshape(y_pred, [-1])

    intersection = tf.reduce_sum(y_true * y_pred)

    return (
        (2.0 * intersection + smooth) /
        (
            tf.reduce_sum(y_true)
            + tf.reduce_sum(y_pred)
            + smooth
        )
    )


def dice_loss(y_true, y_pred):
    return 1.0 - dice_coefficient(y_true, y_pred)


def combined_loss(y_true, y_pred):

    bce = tf.keras.losses.binary_crossentropy(
        y_true,
        y_pred
    )

    return tf.reduce_mean(bce) + dice_loss(
        y_true,
        y_pred
    )
```

```python
IMG_SIZE = 128
BATCH_SIZE = 16
from PIL import Image
import numpy as np
import tensorflow as tf


def load_image_mask_fast(image_path, mask_path):

    image_path = image_path.numpy().decode("utf-8")
    mask_path = mask_path.numpy().decode("utf-8")

    # MRI
    image = Image.open(image_path).convert("RGB")
    image = image.resize((IMG_SIZE, IMG_SIZE))
    image = np.array(image).astype(np.float32) / 255.0

    # Mask
    mask = Image.open(mask_path).convert("L")
    mask = mask.resize(
        (IMG_SIZE, IMG_SIZE),
        Image.Resampling.NEAREST
    )

    mask = np.array(mask).astype(np.float32)

    # 0/255 -> 0/1
    mask = (mask > 0).astype(np.float32)

    mask = np.expand_dims(mask, axis=-1)

    return image, mask
```

```python
def load_image_mask_fast_tf(image_path, mask_path):

    image, mask = tf.py_function(
        func=load_image_mask_fast,
        inp=[image_path, mask_path],
        Tout=[tf.float32, tf.float32]
    )

    image.set_shape([IMG_SIZE, IMG_SIZE, 3])
    mask.set_shape([IMG_SIZE, IMG_SIZE, 1])

    return image, mask
```

```python
def create_fast_dataset(pairs, batch_size=16, shuffle=False):

    image_paths = [p[0] for p in pairs]
    mask_paths = [p[1] for p in pairs]

    dataset = tf.data.Dataset.from_tensor_slices(
        (image_paths, mask_paths)
    )

    dataset = dataset.map(
        load_image_mask_fast_tf,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if shuffle:
        dataset = dataset.shuffle(500)

    dataset = dataset.batch(batch_size)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset


train_ds_fast = create_fast_dataset(
    train_balanced,
    BATCH_SIZE,
    shuffle=True
)

val_ds_fast = create_fast_dataset(
    val_balanced,
    BATCH_SIZE
)

test_ds_fast = create_fast_dataset(
    test_balanced,
    BATCH_SIZE
)

print("Fast datasets created!")
```

```python
from tensorflow.keras import layers, Model


def conv_block(x, filters):
    x = layers.Conv2D(
        filters, 3, padding="same", activation="relu"
    )(x)
    x = layers.BatchNormalization()(x)

    x = layers.Conv2D(
        filters, 3, padding="same", activation="relu"
    )(x)
    x = layers.BatchNormalization()(x)

    return x


def build_unet_128():
    inputs = layers.Input((128, 128, 3))

    # Encoder
    c1 = conv_block(inputs, 16)
    p1 = layers.MaxPooling2D(2)(c1)

    c2 = conv_block(p1, 32)
    p2 = layers.MaxPooling2D(2)(c2)

    c3 = conv_block(p2, 64)
    p3 = layers.MaxPooling2D(2)(c3)

    c4 = conv_block(p3, 128)
    p4 = layers.MaxPooling2D(2)(c4)

    # Bottleneck
    c5 = conv_block(p4, 256)

    # Decoder
    u6 = layers.Conv2DTranspose(
        128, 2, strides=2, padding="same"
    )(c5)
    u6 = layers.concatenate([u6, c4])
    c6 = conv_block(u6, 128)

    u7 = layers.Conv2DTranspose(
        64, 2, strides=2, padding="same"
    )(c6)
    u7 = layers.concatenate([u7, c3])
    c7 = conv_block(u7, 64)

    u8 = layers.Conv2DTranspose(
        32, 2, strides=2, padding="same"
    )(c7)
    u8 = layers.concatenate([u8, c2])
    c8 = conv_block(u8, 32)

    u9 = layers.Conv2DTranspose(
        16, 2, strides=2, padding="same"
    )(c8)
    u9 = layers.concatenate([u9, c1])
    c9 = conv_block(u9, 16)

    outputs = layers.Conv2D(
        1, 1, activation="sigmoid"
    )(c9)

    return Model(inputs, outputs)


model_128 = build_unet_128()

model_128.summary()
```

```python
model_128.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),
    loss=combined_loss,
    metrics=[
        dice_coefficient,
        "accuracy"
    ]
)

print("128x128 U-Net compiled successfully!")
```

```python
import numpy as np

all_dice = []
all_iou = []
all_precision = []
all_recall = []

threshold = 0.5

for images_batch, masks_batch in test_ds:

    preds_batch = model.predict(images_batch, verbose=0)

    preds_batch = preds_batch.squeeze(-1)
    masks_batch = masks_batch.numpy().squeeze(-1)

    for true_mask, pred in zip(masks_batch, preds_batch):

        true = (true_mask > 0.5)
        predicted = (pred >= threshold)

        # Pixels
        intersection = np.logical_and(true, predicted).sum()
        union = np.logical_or(true, predicted).sum()

        true_pixels = true.sum()
        pred_pixels = predicted.sum()

        # Dice
        dice = (2 * intersection) / (
            true_pixels + pred_pixels + 1e-7
        )

        # IoU
        iou = intersection / (union + 1e-7)

        # Precision
        precision = intersection / (
            pred_pixels + 1e-7
        )

        # Recall
        recall = intersection / (
            true_pixels + 1e-7
        )

        all_dice.append(dice)
        all_iou.append(iou)
        all_precision.append(precision)
        all_recall.append(recall)


print("====================================")
print("TEST SET RESULTS")
print("====================================")

print(f"Dice      : {np.mean(all_dice):.4f}")
print(f"IoU       : {np.mean(all_iou):.4f}")
print(f"Precision : {np.mean(all_precision):.4f}")
print(f"Recall    : {np.mean(all_recall):.4f}")
```
